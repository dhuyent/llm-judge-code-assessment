## LABEL 2&3 FILTERING

After re-checking the result of Label 1: if `expected_output`/`actual_output` (test case) is not reasonable or not consistent with the ground-truth label, the `id` is moved from Label 1 $\to$ Label 2.

General definition (with `N` = number of judge attempts: **N = 3** for Stage 1, **N = 2** for Stage 2):

- **Label 1**: all `N/N` correct judge attempts AND manual recheck confirms `pred_reason`/test case is reasonable and consistent with ground truth
- **Label 2**: some but not all correct judge attempts, OR all `N/N` correct but manual recheck marks `pred_reason`/test case as NOT reasonable (failed recheck)
- **Label 3**: `0/N` correct judge attempts

```python
FILTERED_INPUT_FILE  = '/kaggle/input/datasets/dhuyent/classified-stage1/step1_classify1_logic1.csv'
FILTERED_OUTPUT_FILE = 'step1_filter1_logic1.csv'

EXCLUDE_COLUMNS = [
    'llm_model', 'prompt_strategy', 'pred_status', 'pred_input',
    'pred_actual_output', 'pred_expected_output', 'pred_reason',
    'match', 'label', 'filtered_label'
]

df_stage1 = pd.read_csv(FILTERED_INPUT_FILE)

df_label_2_3 = df_stage1[df_stage1['filtered_label'].isin([2, 3])]
df_label_2_3 = df_label_2_3.drop(columns=EXCLUDE_COLUMNS).reset_index(drop=True)

df_label_2_3.to_csv(FILTERED_OUTPUT_FILE, index=False, encoding='utf-8-sig')
print(f"Saved {len(df_label_2_3)} rows (filtered_label 2 & 3) to {FILTERED_OUTPUT_FILE}")
```
