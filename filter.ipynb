## LABEL 2&3 FILTERING

After re-checking result of Label 1: if `expected_output`/`actual_output` (test case) is not reasonable and consistent with the ground-truth label, the `id` is moved from Label 1 $\to$ Label 2:

- Label 1: 3/3 correct judge attempts AND manual recheck confirms `pred_reason`/test case is reasonable and consistent with ground truth
- Label 2: 1/3 or 2/3 correct judge attempts OR 3/3 correct judge attempts but manual recheck marks `pred_reason`/test case as NOT reasonable (failed recheck)
- Label 3: 0/3 correct judge attempt

```python
FILTERED_INPUT_FILE  = '/kaggle/input/datasets/dhuyent/classified-stage1/step1_classify1_logic1.csv'
FILTERED_OUTPUT_FILE = 'step1_filter1_logic1.csv'

EXCLUDE_COLUMNS = [
    'llm_model', 'prompt_strategy', 'pred_status', 'pred_input',
    'pred_actual_output', 'pred_expected_output', 'pred_reason',
    'comp.', 'match', 'label'
]

df_stage1 = pd.read_csv(FILTERED_INPUT_FILE)

df_label_2_3 = df_stage1[df_stage1['filtered_label'].isin([2, 3])]
df_label_2_3 = df_label_2_3.drop(columns=EXCLUDE_COLUMNS).reset_index(drop=True)

df_label_2_3.to_csv(FILTERED_OUTPUT_FILE, index=False, encoding='utf-8-sig')
print(f"Saved {len(df_label_2_3)} rows (filtered_label 2 & 3) to {FILTERED_OUTPUT_FILE}")
```
